```python
from pymongo import MongoClient
import csv
from google.colab import drive
from io import StringIO
import requests
from pprint import pprint
```

```python
# the connection to the mongodb database and subsequent collections
client = MongoClient("mongodb+srv://shaneob4123_db_user:77JwMek0qmnNQl3W@northstar-app.qs6x62p.mongodb.net/?appName=northstar-app")
db = client["north-star"]
app_collection = db["events-history"]
deliveries_collection = db["deliveries"]
orders_collection = db["Orders"]
customers_collection = db["Customers"]
```

```python
# Deletes any data already stored in the database collections so that it doesn't duplicate when running again
deliveries_collection.delete_many({})
orders_collection.delete_many({})
customers_collection.delete_many({})
```

```python
# import cleaned data into the database

# github raw file for cleaned data
deliveries = "https://raw.githubusercontent.com/milkmaN4123/Data_bases_Cw_northstar/refs/heads/clean/clean/clean_deliveries.csv"
orders = "https://raw.githubusercontent.com/milkmaN4123/Data_bases_Cw_northstar/refs/heads/clean/clean/clean_orders.csv"
customers = "https://raw.githubusercontent.com/milkmaN4123/Data_bases_Cw_northstar/refs/heads/clean/clean/clean_customers.csv"
```

```python
response_deliveries = requests.get(deliveries)
response_deliveries.raise_for_status()

deliveries_text = StringIO(response_deliveries.text)
reader_deliveries = csv.DictReader(deliveries_text)
rows_deliveries = list(reader_deliveries)

if rows_deliveries:
    deliveries_collection.insert_many(rows_deliveries)
    print("Imported", len(rows_deliveries), "rows")
else:
    print("No rows found")
```

```python
response_customers = requests.get(customers)
response_customers.raise_for_status()

customers_text = StringIO(response_customers.text)
reader_customers = csv.DictReader(customers_text)
rows_customers = list(reader_customers)

if rows_customers:
    customers_collection.insert_many(rows_customers)
    print("Imported", len(rows_customers), "rows")
else:
    print("No rows found")
```

```python
response_orders = requests.get(orders)
response_orders.raise_for_status()

orders_text = StringIO(response_orders.text)
reader_orders = csv.DictReader(orders_text)
rows_orders = list(reader_orders)

if rows_orders:
    orders_collection.insert_many(rows_orders)
    print("Imported", len(rows_orders), "rows")
else:
    print("No rows found")
```

```python
# create an index for order_id
# by adding an index it helps with faster querying

app_collection.create_index("order_id",unique = True)
```

```python
# find first 10 rows of app collection
app_documents = app_collection.find().limit(10)
for app_doc in app_documents:
    print(app_doc)
```

```python
# find documents with an event type of "payment retry"

for doc in app_collection.find({"events.event_type": "payment_retry"}):
    pprint({
        "order_id": doc["order_id"],
        "customer_id": doc["customer_id"],
        "event_count": doc["summary"]["event_count"]
    })
```

```python
# find documents with failed events

for doc in app_collection.find({"summary.failed_event_count": {"$gt": 0}}).limit(10):
    pprint({
        "order_id": doc["order_id"],
        "failed_event_count": doc["summary"]["failed_event_count"]
    })
```
